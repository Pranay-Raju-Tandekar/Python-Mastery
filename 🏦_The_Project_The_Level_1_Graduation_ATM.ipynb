🟡 Level 2, Step 2: Dictionaries (dict)

```
# Creating a user profile dictionary
user_profile = {
    "name": "Rahul",
    "role": "Data Analyst",
    "experience_years": 3,
    "is_certified": True
}

# Looking up data using the Key
print(user_profile["name"])  # Prints: Rahul
print(user_profile["role"])  # Prints: Data Analyst

# Adding a brand new Key-Value pair
user_profile["location"] = "India"

# Modifying an existing value
user_profile["experience_years"] = 4
```

### Dictionaries Practice

1 Create a dictionary called product with these three key-value pairs:

"title" -> "Laptop"

"price" -> 45000

"stock" -> 10

2 Write a line of code to update the price to 42000 (simulating a discount).

3 Write a line of code to add a new key called "brand" with the value "Dell".

4 Use a print() statement to show the entire dictionary to verify your changes.

```
# 1. Setting up the initial dictionary
product = {
    "title": "Laptop",
    "price": 45000,
    "stock": 10
}

# 2. Corrected: No extra space inside "price"
product["price"] = 42000

# 3. Corrected: Matching double quotes around "brand"
product["brand"] = "Dell"

# 4. Print the final result to check your work
print(product)
```
