## Level 1: Graduation Project.

```python
# 1. Variables & Data Types (Initial Setup)
account_balance = 5000.0  # float
pin = 1234               # int
user_name = "Rahul"       # str
is_running = True         # bool

print("=== WELCOME TO THE PYTHON ATM ===")

# 2. The While Loop (Keeps the ATM running until the user chooses to exit)
while is_running == True:
    print("\n1. Check Balance\n2. Withdraw Money\n3. Deposit Money\n4. Exit")

    # Taking user input (Advanced tip: input() always takes text, so we convert it to an int)
    choice = int(input("Choose an option (1-4): "))

    # 3. Control Flow (Decision Making)
    if choice == 1:
        # Task: Print a nice message showing the user_name and their account_balance
        print(f"Hello {user_name}, your balance is Rs. {account_balance}")

    elif choice == 2:
        amount = float(input("Enter amount to withdraw: "))

        # Task: Use an 'if' statement to check if 'amount' is less than or equal to 'account_balance'
        # If they have enough money: subtract the amount from account_balance and print "Withdrawal successful!"
        # Else: print "Insufficient funds!"
        if amount <= account_balance:
            account_balance = account_balance - amount
            print("Withdrawal successful!")
        else:
            print("Insufficient funds!")

    elif choice == 3:
        deposit_amount = float(input("Enter amount to deposit: "))
        # Task: Add the deposit_amount to the account_balance and print the new balance
        account_balance = account_balance + deposit_amount
        print(f"Deposited successfully! New balance: Rs. {account_balance}")

    elif choice == 4:
        # Task: Change the boolean variable that controls the loop so the loop stops running
        print("Thank you for using the Python ATM. Goodbye!")
        is_running = False

    else:
        print("Invalid choice! Please select a number between 1 and 4.")
```

🏛️ Level 3 Graduation: The Bulletproof Currency Converter

```python
def convert_inr_to_usd(inr_amount):
    """Converts INR to USD assuming 1 USD = 83 INR."""
    # Task: Divide inr_amount by 83 and return the result
    usd_value = inr_amount / 83
    return usd_value

print("--- SECURE EXCHANGE TERMINAL ---")

# We use a try block because input() and calculations can easily crash if a user types text instead of a number
try:
    # 1. Take raw input from the user
    user_input = input("Enter amount in INR to convert: ")

    # 2. Convert the input text string into a clean float number
    clean_inr = float(user_input)

    # 3. Call your function and pass the clean number into it
    final_usd = convert_inr_to_usd(clean_inr)

    # 4. Print the final calculated amount using an F-string (rounded to 2 decimal places)
    print(f"Total Exchange: ${final_usd:.2f} USD")

except ValueError:
    # This block intercepts the error if a user types something like "abc" instead of numbers
    print("❌ Input Error: Please enter a valid numerical value. Transaction canceled.")

print("--------------------------------")
print("Terminal status: Secure & operational.")
```

```python
# 1. Import the library
import numpy as np

# 2. Create a standard list of monthly salaries in INR
salaries_list = [50000, 65000, 72000, 48000]

# 3. Convert that list into a high-performance NumPy array
salaries_array = np.array(salaries_list)

# 4. TASK: Give everyone a 10% bonus!
# Multiply the salaries_array by 1.10 directly without using any loops
bonus_salaries = salaries_array * 1.10

# 5. Print the new array to see the vectorized calculation magic
print(bonus_salaries)
print(f"{bonus_salaries.mean()}")
print(f"{bonus_salaries.max()}")
print(bonus_salaries > 60000)

high_salaries = bonus_salaries[bonus_salaries > 60000]
print(high_salaries)
```

# Task 3: Filter the array to find only high-efficiency days (production > 500)

```python
import numpy as np


daily_production = np.array([450, 520, 380, 610, 490, 580, 310])

print("--- FACTORY PERFORMANCE REPORT ---")


avg_prod = daily_production.mean()
print(f"Average Daily Production: {avg_prod:.1f} units")


peak_prod = daily_production.max()
print(f"Peak Production: {peak_prod} units")


high_efficiency_days = daily_production[daily_production > 500]
print("High-Efficiency Days Production:", high_efficiency_days)
```

```python
import pandas as pd

# 1. Expanded corporate enterprise dataset
company_data = {
    "Employee": ["Rahul", "Priya", "Amit", "Suresh", "Neha"],
    "Department": ["Data", "DevOps", "Design", "Data", "DevOps"],
    "Salary": [55000, 71500, 79200, 62000, 85000],
    "Performance_Score": [4.5, 4.8, 3.9, 4.2, 4.9]
}

# 2. Convert to DataFrame
df = pd.DataFrame(company_data)

print("============ ENTERPRISE HR DASHBOARD ============")
print(df)
print("=================================================")

# Task 1: Calculate the Total Monthly Payroll using .sum() on the "Salary" column
total_payroll = df["Salary"].sum()
print(f"Total Monthly Payroll Expense : Rs. {total_payroll:,}")

# Task 2: Calculate the Average Performance Score using .mean() on the "Performance_Score" column
avg_performance = df["Performance_Score"].mean()
print(f"Company Average Performance   : {avg_performance:.2f} / 5.0")
print("=================================================")
```
