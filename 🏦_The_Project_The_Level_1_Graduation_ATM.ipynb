## Level 1: Graduation Project.

```python
# 1. Variables & Data Types (Initial Setup)
account_balance = 5000.0  # float
pin = 1234               # int
user_name = "Rahul"       # str
is_running = True         # bool

print("=== WELCOME TO THE PYTHON ATM ===")

# 2. The While Loop (Keeps the ATM running until the user chooses to exit)
while is_running == True:
    print("\n1. Check Balance\n2. Withdraw Money\n3. Deposit Money\n4. Exit")

    # Taking user input (Advanced tip: input() always takes text, so we convert it to an int)
    choice = int(input("Choose an option (1-4): "))

    # 3. Control Flow (Decision Making)
    if choice == 1:
        # Task: Print a nice message showing the user_name and their account_balance
        print(f"Hello {user_name}, your balance is Rs. {account_balance}")

    elif choice == 2:
        amount = float(input("Enter amount to withdraw: "))

        # Task: Use an 'if' statement to check if 'amount' is less than or equal to 'account_balance'
        # If they have enough money: subtract the amount from account_balance and print "Withdrawal successful!"
        # Else: print "Insufficient funds!"
        if amount <= account_balance:
            account_balance = account_balance - amount
            print("Withdrawal successful!")
        else:
            print("Insufficient funds!")

    elif choice == 3:
        deposit_amount = float(input("Enter amount to deposit: "))
        # Task: Add the deposit_amount to the account_balance and print the new balance
        account_balance = account_balance + deposit_amount
        print(f"Deposited successfully! New balance: Rs. {account_balance}")

    elif choice == 4:
        # Task: Change the boolean variable that controls the loop so the loop stops running
        print("Thank you for using the Python ATM. Goodbye!")
        is_running = False

    else:
        print("Invalid choice! Please select a number between 1 and 4.")
```
